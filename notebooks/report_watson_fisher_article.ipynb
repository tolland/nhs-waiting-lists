```python
# Get the sqlite database path

import nhs_waiting_lists as nhs
from nhs_waiting_lists.constants import ALL_TREATMENT_CODES
```

> Over the two previous complete financial years (from April 2023 to March this year), there have been on average 244,578 unreported removals a month (unadjusted for working days), and 7,566,344 incomplete pathways (making up the reported waiting list) a month. This means that, on average, unreported removals equate to around 3% of the size of the waiting list.

Source: <https://www.nuffieldtrust.org.uk/news-item/why-is-the-planned-care-waiting-list-coming-down-and-what-does-the-data-really-tell-us?dm_i=21A8,909FB,1TOIL8,11MABM,1>


```python
# get the consolidated data corresponding to the previously published estimates

consolidated_df = nhs.get_consolidated_df2(
    "2023-04",
    "2025-03",
    [],
    ALL_TREATMENT_CODES,
    # ["C_999"],
)

consolidated_df.head()
```

```python
consolidated_c_999_df = nhs.get_consolidated_df2(
    "2023-04",
    "2025-03",
    [],
    ["C_999"],
)

consolidated_c_999_df.head()
```

```python
import pandas as pd

other_result: pd.DataFrame = consolidated_df.groupby(['period']).sum()

# need to calcualte the pct after the aggregation
other_result["unexplained_pct"] = other_result["untreated"] / other_result["total_treatable"]
print(f"untreated mean: {other_result['untreated'].mean()}")
print(f"incomplete mean: {other_result['incomplete'].mean()}")
print(f"incomplete mean: {other_result['incomplete'].mean()}")
```

```python
other_result_c_999: pd.DataFrame = consolidated_c_999_df.groupby(['period']).sum()

# need to calcualte the pct after the aggregation
other_result_c_999["unexplained_pct"] = other_result_c_999["untreated"] / other_result_c_999["total_treatable"]
print(f"untreated mean: {other_result_c_999['untreated'].mean():0.2f}")
print(f"incomplete mean: {other_result_c_999['incomplete'].mean():0.2f}")
```

```python
for i in range(5):
    start_period = f"2023-{2+i:>02}"
    end_period = f"2025-{1+i:>02}"
    print(f"from {start_period} to {end_period}")

    df_raw = nhs.get_consolidated_df2(
        start_period,
        end_period,
        [], # give all providers
        # ["C_999"], # use 'total' treatment code
        ALL_TREATMENT_CODES,
    )
    other_result_c_999: pd.DataFrame = df_raw.groupby(['period']).sum()
    print(f"untreated mean: {other_result_c_999['untreated'].mean():0.2f}")
    print(f"incomplete mean: {other_result_c_999['incomplete'].mean():0.2f}")
    print(other_result_c_999['untreated'].mean()/other_result_c_999['incomplete'].mean())
    print("")
```
